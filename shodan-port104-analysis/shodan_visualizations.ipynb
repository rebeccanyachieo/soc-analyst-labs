```python
import json
import pandas as pd
import matplotlib.pyplot as plt
```

```python
with open("shodan_port104_20260724_222242Z.json", "r") as f:
    data = json.load(f)
    
    df = pd.DataFrame(data)
    
    public_df = df.drop(columns=["ip"])
    
```

```python
country_counts = public_df["country"].value_counts()
country_counts.head(10)
```

```python
country_counts.plot(kind="barh")

plt.title("Port 104 Shodan Results by Country")
plt.xlabel("Number of Hosts")
plt.ylabel("Country")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
org_counts = public_df["organization"].value_counts().head(10)

org_counts.plot(kind="barh")

plt.title("Top Organizations Associated with Port 104 Results")
plt.xlabel("Number of Hosts")
plt.ylabel("Organization")
plt.tight_layout()
plt.show()
```

```python
org_country = (
    public_df.groupby(["country", "organization"])
        .size()
        .reset_index(name="count")
        .sort_values("count", ascending=False)
)
org_country.head(15)
```
